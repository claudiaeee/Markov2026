```python
import numpy as np
```

```python
T = 2*10**5
period = 5
rate_s = 1.2 * period
rate_b = 0.8 * period
rate_tot = 2 * period
```

```python
Ntot = np.random.poisson(T*rate_tot)
times = np.random.rand(Ntot) * T
times = np.sort(times)
probs = np.random.rand(Ntot)
times_b1 = times[probs < 0.4]
times_s1 = times[probs >= 0.4]
Nb1 = np.bincount(times_b1.astype(int))
Ns1 = np.bincount(times_s1.astype(int))
N1 = np.bincount(times.astype(int))
print(times_s1)
print(Nb1)
```

```python
Ns = np.random.poisson(T*rate_s)
times_s2 = np.random.rand(Ns) * T
times_s2 = np.sort(times_s2)
Ns2 = np.bincount(times_s2.astype(int))
Nb = np.random.poisson(T*rate_b)
times_b2 = np.random.rand(Nb) * T
times_b2 = np.sort(times_b2)
Nb2 = np.bincount(times_b2.astype(int))
N2 = np.resize(Ns2, len(Nb2))+Nb2

meanb1 = np.mean(Nb1)
means1 = np.mean(Ns1)
meanb2 = np.mean(Nb2)
means2 = np.mean(Ns2)
varb1 = np.var(Nb1)
vars1 = np.var(Ns1)
varb2 = np.var(Nb2)
vars2 = np.var(Ns2)
print("Variance for background photon count using 1st simluation method: " + str(varb1))
print("Variance for background photon count using 2nd simluation method: " + str(varb2))
print("Variance for star photon count using 1st simluation method: " + str(vars1))
print("Variance for star photon count using 2nd simluation method: " + str(vars2))
print("Mean for background photon count using 1st simulation method: " + str(meanb1))
print("Mean for background photon count using 2nd simulation method: " + str(meanb2))
print("Mean for star photon count using 1st simulation method: " + str(means1))
print("Mean for star photon count using 2nd simulation method: " + str(means2))
print("Correlation between star and background counts, sim1: " + str(np.corrcoef(Ns1, Nb1)[0,1]))
print("Correlation between star and background counts, sim2: " + str(np.corrcoef(Ns2, Nb2)[0,1]))
```
